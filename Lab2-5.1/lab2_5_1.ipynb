# 5.1.1. MidSquare

```python
import pandas as pd

def midsquare_generator(seed, n):
    """
    Genera una secuencia de números pseudoaleatorios usando el método
    de MidSquare (Cuadrados Medios).
    """
    d = len(str(seed))
    X = seed
    X_list = []
    u_list = []

    for _ in range(n):
        X2 = X ** 2
        X2_str = str(X2).zfill(2 * d)
        start = (2 * d - d) // 2
        mid_str = X2_str[start:start + d]
        X = int(mid_str)

        X_list.append(X)
        u_list.append(X / (10 ** d))

    return X_list, u_list


# Semillas del punto 5.1.1
seeds = {
    'Semilla a (1931)': 1931,
    'Semilla b (675248)': 675248,
    'Semilla c (4567)': 4567,
    'Semilla d (12122)': 12122,
    'Semilla e (7179345)': 7179345
}

n = 100

# Construimos un diccionario con los u_i de cada semilla
data = {'Iteración': list(range(1, n + 1))}

for nombre, seed in seeds.items():
    _, u_seq = midsquare_generator(seed, n)
    data[nombre] = [round(u, 4) for u in u_seq]

# Tabla única con todas las secuencias
tabla_midsquare = pd.DataFrame(data)
tabla_midsquare
```

# 5.1.2. Generador congruencial mixto

```python
import pandas as pd

def lcg_generator(X0, a, c, m, n):
    X = X0
    X_list = []
    u_list = []

    for _ in range(n):
        X = (a * X + c) % m
        X_list.append(X)
        u_list.append(X / m)

    return X_list, u_list


# Parámetros
parametros = {
    'a) X0=127,a=115,c=0,m=128':        {'X0': 127,      'a': 115,          'c': 0,  'm': 128},
    'b) X0=43,a=25,c=5,m=48':           {'X0': 43,       'a': 25,           'c': 5,  'm': 48},
    'c) X0=13,a=16807,c=0,m=2^31-1':    {'X0': 13,       'a': 16807,        'c': 0,  'm': 2**31 - 1},
    'd) X0=29022024,a=25214903917,c=11,m=2^48': {'X0': 29022024, 'a': 25214903917, 'c': 11, 'm': 2**48},
    'e) X0=3,a=5,c=7,m=200':            {'X0': 3,        'a': 5,            'c': 7,  'm': 200},
}

n = 100

# Construimos la tabla
data = {'Iteración': list(range(1, n + 1))}
resultados = {}  # guardamos también los X_i por si se necesitan después (ciclo, etc.)

for nombre, p in parametros.items():
    X_seq, u_seq = lcg_generator(p['X0'], p['a'], p['c'], p['m'], n)
    resultados[nombre] = {'X': X_seq, 'u': u_seq, 'params': p}
    data[nombre] = [round(u, 4) for u in u_seq]

tabla_glc = pd.DataFrame(data)


tabla_glc
```

# 5.1.3. Otros generadores de números aleatorios
**Generador Shift-Register** y
**Lagged-Fibonnaci**

```python
import pandas as pd

def lagged_fibonacci_generator(seeds, j, k, m, n, op='+'):
    assert len(seeds) >= k, "Se necesitan al menos k semillas"
    X = list(seeds)

    for i in range(k, k + n):
        if op == '+':
            X.append((X[i - j] + X[i - k]) % m)
        else:
            X.append((X[i - j] * X[i - k]) % m)

    X_generados = X[k:k + n]
    u_list = [x / m for x in X_generados]
    return X_generados, u_list


# Parámetros de ejemplo
j_lf, k_lf, m_lf = 5, 17, 2**31
seeds_lf = [1931, 675248, 4567, 12122, 7179345,
            2391, 8842, 173, 9284, 55321,
            883, 20194, 671, 3348, 9021,
            4471, 12]

n = 100
X_lf, u_lf = lagged_fibonacci_generator(seeds_lf, j_lf, k_lf, m_lf, n, op='+')

def shift_register_generator(seed_bits, p, q, n_bits_out, n):
    bits = list(seed_bits)
    total_bits_needed = n * n_bits_out

    while len(bits) < len(seed_bits) + total_bits_needed:
        i = len(bits)
        nuevo_bit = bits[i - p] ^ bits[i - q]
        bits.append(nuevo_bit)

    # Descartamos los bits semilla y agrupamos el resto en bloques
    bits_generados = bits[len(seed_bits):]
    X_list = []
    for i in range(n):
        bloque = bits_generados[i * n_bits_out:(i + 1) * n_bits_out]
        valor = int(''.join(map(str, bloque)), 2)
        X_list.append(valor)

    max_val = 2 ** n_bits_out - 1
    u_list = [x / max_val for x in X_list]
    return X_list, u_list


# Parámetros de ejemplo para Shift-Register
p_sr, q_sr = 7, 3
n_bits_out = 10
seed_bits_sr = [1, 0, 1, 1, 0, 0, 1]  # semilla binaria, longitud = p_sr


X_sr, u_sr = shift_register_generator(seed_bits_sr, p_sr, q_sr, n_bits_out, n)


data = {
    'Iteración': list(range(1, n + 1)),
    'Lagged-Fibonacci (u_i)': [round(u, 4) for u in u_lf],
    'Shift-Register (u_i)': [round(u, 4) for u in u_sr],
}

tabla_531 = pd.DataFrame(data)

tabla_531
```

# 5.1.4. Tamaño de ciclo, pruebas de uniformidad y aleatoriedad

```python
import pandas as pd
import numpy as np
from scipy import stats

# =========================================================
# 1. GENERADORES (versión "iterador" para poder detectar ciclos)
# =========================================================

def midsquare_gen(seed):
    d = len(str(seed))
    X = seed
    while True:
        X2 = X ** 2
        X2_str = str(X2).zfill(2 * d)
        start = (2 * d - d) // 2
        X = int(X2_str[start:start + d])
        yield X, d  # devolvemos también d para poder normalizar

def lcg_gen(X0, a, c, m):
    X = X0
    while True:
        X = (a * X + c) % m
        yield X

def lagged_fibonacci_gen(seeds, j, k, m, op='+'):
    state = list(seeds)
    i = k
    while True:
        val = (state[i - j] + state[i - k]) % m if op == '+' else (state[i - j] * state[i - k]) % m
        state.append(val)
        i += 1
        yield val

def shift_register_gen(seed_bits, p, q, n_bits_out):
    bits = list(seed_bits)
    i = len(bits)
    while True:
        block = []
        for _ in range(n_bits_out):
            nuevo = bits[i - p] ^ bits[i - q]
            bits.append(nuevo)
            block.append(nuevo)
            i += 1
        yield int(''.join(map(str, block)), 2)


# =========================================================
# 2. RE-GENERAMOS LAS 12 SECUENCIAS (n = 100) DE 5.1.1, 5.1.2, 5.1.3
# =========================================================
n = 100
secuencias = {}

# --- 5.1.1 MidSquare ---
seeds_ms = {'MS-a (1931)': 1931, 'MS-b (675248)': 675248, 'MS-c (4567)': 4567,
            'MS-d (12122)': 12122, 'MS-e (7179345)': 7179345}

for nombre, seed in seeds_ms.items():
    d = len(str(seed))
    g = midsquare_gen(seed)
    X_list = []
    for _ in range(n):
        X, _ = next(g)
        X_list.append(X)
    u_list = [x / (10 ** d) for x in X_list]
    secuencias[nombre] = {'u': u_list, 'X': X_list,
                           'factory': lambda s=seed: (v for v, _ in midsquare_gen(s))}

# --- 5.1.2 GLC ---
params_glc = {
    'GLC-a': {'X0': 127, 'a': 115, 'c': 0, 'm': 128},
    'GLC-b': {'X0': 43, 'a': 25, 'c': 5, 'm': 48},
    'GLC-c': {'X0': 13, 'a': 16807, 'c': 0, 'm': 2**31 - 1},
    'GLC-d': {'X0': 29022024, 'a': 25214903917, 'c': 11, 'm': 2**48},
    'GLC-e': {'X0': 3, 'a': 5, 'c': 7, 'm': 200},
}

for nombre, p in params_glc.items():
    g = lcg_gen(p['X0'], p['a'], p['c'], p['m'])
    X_list = [next(g) for _ in range(n)]
    u_list = [x / p['m'] for x in X_list]
    secuencias[nombre] = {'u': u_list, 'X': X_list,
                           'factory': lambda p=p: lcg_gen(p['X0'], p['a'], p['c'], p['m'])}

# --- 5.1.3 Lagged-Fibonacci ---
j_lf, k_lf, m_lf = 5, 17, 2**31
seeds_lf = [1931, 675248, 4567, 12122, 7179345, 2391, 8842, 173, 9284, 55321,
            883, 20194, 671, 3348, 9021, 4471, 12]
g = lagged_fibonacci_gen(seeds_lf, j_lf, k_lf, m_lf, op='+')
X_lf = [next(g) for _ in range(n)]
u_lf = [x / m_lf for x in X_lf]
secuencias['Lagged-Fibonacci'] = {'u': u_lf, 'X': X_lf,
                                   'factory': lambda: lagged_fibonacci_gen(seeds_lf, j_lf, k_lf, m_lf, op='+')}

# --- 5.1.3 Shift-Register ---
p_sr, q_sr, n_bits_out = 7, 3, 10
seed_bits_sr = [1, 0, 1, 1, 0, 0, 1]
g = shift_register_gen(seed_bits_sr, p_sr, q_sr, n_bits_out)
X_sr = [next(g) for _ in range(n)]
max_val_sr = 2 ** n_bits_out - 1
u_sr = [x / max_val_sr for x in X_sr]
secuencias['Shift-Register'] = {'u': u_sr, 'X': X_sr,
                                 'factory': lambda: shift_register_gen(seed_bits_sr, p_sr, q_sr, n_bits_out)}


# =========================================================
# 3. (a) TAMAÑO DE CICLO O PERIODO
# =========================================================
def find_cycle(gen_iterator, max_iter=100_000):
    """
    Detecta el tamaño de ciclo generando valores hasta encontrar una
    repetición. Retorna (longitud_cola, longitud_ciclo) o (None, None)
    si no se detecta dentro de max_iter iteraciones (generador de periodo
    muy largo -> "bueno" en ese sentido).
    """
    seen = {}
    idx = 0
    for val in gen_iterator:
        if val in seen:
            return seen[val], idx - seen[val]
        seen[val] = idx
        idx += 1
        if idx >= max_iter:
            return None, None
    return None, None


# =========================================================
# 4. (b) PRUEBA DE UNIFORMIDAD -> CHI-CUADRADO
# =========================================================
def chi_square_uniformity_test(u_list, k=10, alpha=0.05):
    """
    Divide [0,1) en k intervalos y compara frecuencias observadas vs
    esperadas (n/k) usando la prueba Chi-cuadrado de bondad de ajuste.
    """
    n = len(u_list)
    observed, _ = np.histogram(u_list, bins=k, range=(0, 1))
    expected = np.full(k, n / k)
    chi2_stat = np.sum((observed - expected) ** 2 / expected)
    df = k - 1
    chi2_critico = stats.chi2.ppf(1 - alpha, df)
    p_value = 1 - stats.chi2.cdf(chi2_stat, df)
    es_uniforme = chi2_stat <= chi2_critico
    return chi2_stat, chi2_critico, p_value, es_uniforme


# =========================================================
# 5. (c) PRUEBA DE ALEATORIEDAD -> TEST DE RACHAS (arriba/abajo de la mediana)
# =========================================================
def runs_test(u_list, alpha=0.05):
    """
    Test de rachas por encima/por debajo de la mediana (0.5, ya que u_i ~ U(0,1)).
    Calcula el estadístico Z y lo compara con el valor crítico normal.
    """
    n = len(u_list)
    signos = ['+' if u >= 0.5 else '-' for u in u_list]

    n1 = signos.count('+')
    n2 = signos.count('-')

    # Contar rachas (cambios de signo + 1)
    R = 1
    for i in range(1, n):
        if signos[i] != signos[i - 1]:
            R += 1

    if n1 == 0 or n2 == 0:
        return R, None, None, False  # caso degenerado

    mu_R = (2 * n1 * n2) / (n1 + n2) + 1
    sigma2_R = (2 * n1 * n2 * (2 * n1 * n2 - n1 - n2)) / (((n1 + n2) ** 2) * (n1 + n2 - 1))
    sigma_R = np.sqrt(sigma2_R)

    Z = (R - mu_R) / sigma_R
    Z_critico = stats.norm.ppf(1 - alpha / 2)  # prueba de dos colas
    es_aleatorio = abs(Z) <= Z_critico

    return R, Z, Z_critico, es_aleatorio


# =========================================================
# 6. APLICAR LAS 3 PRUEBAS A CADA UNA DE LAS 12 SECUENCIAS
# =========================================================
resumen = []

for nombre, datos in secuencias.items():
    u_list = datos['u']

    # (a) Ciclo -- se genera un stream nuevo e independiente para no gastar el u_list ya usado
    tail, cycle_len = find_cycle(datos['factory'](), max_iter=100_000)
    if cycle_len is None:
        ciclo_str = "> 100,000 (no detectado)"
    else:
        ciclo_str = f"{cycle_len} (cola={tail})"

    # (b) Uniformidad
    chi2_stat, chi2_crit, p_val, es_uniforme = chi_square_uniformity_test(u_list, k=10, alpha=0.05)

    # (c) Aleatoriedad (rachas)
    R, Z, Z_crit, es_aleatorio = runs_test(u_list, alpha=0.05)

    resumen.append({
        'Generador': nombre,
        'Tamaño de ciclo': ciclo_str,
        'Chi² calculado': round(chi2_stat, 3),
        'Chi² crítico': round(chi2_crit, 3),
        'p-valor': round(p_val, 4),
        '¿Uniforme?': 'Sí' if es_uniforme else 'No',
        'Nº Rachas (R)': R,
        'Z calculado': round(Z, 3) if Z is not None else 'N/A',
        'Z crítico': round(Z_crit, 3) if Z_crit is not None else 'N/A',
        '¿Aleatorio?': 'Sí' if es_aleatorio else 'No',
    })

tabla_resumen = pd.DataFrame(resumen)

tabla_resumen
```
