# Práctica 01 - Simulación Ad Hoc
## Intervalos de confianza de las medidas de desempeño

Este notebook sigue el mismo esquema del ejemplo del profesor (`Confidence_intervals.ipynb`): para cada medida se listan los valores por corrida y luego se despliega, en formato LaTeX, la media, la desviación estándar, el valor $t$, la media anchura ($h$) y el intervalo de confianza.

Se calculan por separado los dos escenarios pedidos en la guía:

- **Puntos 5.1 a 5.4:** simulación con **20 clientes**
- **Punto 5.5:** simulación con **200 clientes**

```python
import numpy as np
from scipy import stats
from IPython.display import display, Markdown
```

## Funciones auxiliares

Mismo procedimiento del ejemplo del profesor, generalizado para poder reutilizarlo con cualquier medida de desempeño y cualquier nivel de confianza.

```python
def mostrar_datos(nombre_medida, M):
    """Imprime la tabla Rep. / Valor para una medida de desempeño, igual que el ejemplo del profesor."""
    display(Markdown(f"### {nombre_medida}"))
    print(M)
    reps = len(M)
    print("Rep.\tValor")
    for i in range(0, len(M)):
        print("%d\t%g" % (i + 1, M[i]), sep=' ', end='\n')
    return reps


def mostrar_ic(M, alpha):
    """Calcula y despliega en LaTeX la media, desviacion, t, half-width e IC, igual que el ejemplo del profesor."""
    reps = len(M)
    confidence = (1 - alpha) * 100

    mean = np.mean(M)
    var = np.var(M, ddof=1)
    desv = np.sqrt(var)

    tval = stats.t.ppf(1 - (alpha / 2), reps - 1)
    hval = tval * (desv / np.sqrt(reps))

    display(Markdown(
        rf"""
${confidence}\%$ confianza

$\bar{{X}} = {mean:.4f}$

$S = {desv:.4f}$

$t_{{n-1,1-\frac{{\alpha}}{2}}} = t_{{{reps-1},{1-alpha/2}}} = {tval:.4f}$

$h = {hval:.4f}$

Intervalo de confianza : $C.I. = (\bar{{X}}-h , \bar{{X}} + h) = ({mean-hval:.4f}, {mean+hval:.4f})$

"""))
```

# 5.1 a 5.4 — Simulación con 20 clientes

Datos de las 10 corridas (runs) registradas para el escenario de 20 clientes.

```python
data_20 = {
    "Average time in system": [3.95, 4.75, 4.4, 4.15, 3.8, 4.1, 3.85, 5.55, 6, 3.95],
    "Percent idle time": [0.38, 0.36, 0.34, 0.35, 0.46, 0.32, 0.47, 0.17, 0.20, 0.36],
    "Average waiting time per customer": [0.65, 0.95, 0.8, 0.95, 0.4, 0.55, 0.85, 1.7, 1.95, 0.25],
    "Fraction having to wait": [0.30, 0.45, 0.40, 0.40, 0.15, 0.20, 0.30, 0.55, 0.55, 0.25],
    "Average waiting time of those who waited": [2.166666667, 2.111111111, 2.0, 2.375, 2.666666667,
                                                  2.75, 2.833333333, 3.090909091, 3.545454545, 1.0],
}
```

### 5.4 — Intervalos de confianza (20 clientes)

Para cada medida se despliega primero la tabla de valores por corrida y luego el intervalo de confianza al 95 % y al 99 %.

```python
for nombre_medida, M in data_20.items():
    mostrar_datos(nombre_medida, M)
    mostrar_ic(M, alpha=0.05)
    mostrar_ic(M, alpha=0.01)
```

### Intervalos de confianza (200 clientes)

Mismo procedimiento del punto 5.4, repetido aquí para el escenario de 200 clientes.

```python
for nombre_medida, M in data_200.items():
    mostrar_datos(nombre_medida, M)
    mostrar_ic(M, alpha=0.05)
    mostrar_ic(M, alpha=0.01)
```

## Comparación final: ¿cuál combinación tiene la menor desviación estándar y el menor rango de IC?

Se arma una tabla resumen con las dos simulaciones, las cinco medidas y los dos niveles de confianza, para identificar directamente la combinación con menor `S` y menor rango de intervalo.

```python
import pandas as pd

resumen = []
for nombre_dataset, dataset in [("20 clientes", data_20), ("200 clientes", data_200)]:
    for nombre_medida, M in dataset.items():
        for alpha in [0.05, 0.01]:
            reps = len(M)
            mean = np.mean(M)
            desv = np.std(M, ddof=1)
            tval = stats.t.ppf(1 - (alpha / 2), reps - 1)
            hval = tval * (desv / np.sqrt(reps))
            resumen.append({
                "Clientes": nombre_dataset,
                "Medida": nombre_medida,
                "Confianza": f"{int((1-alpha)*100)}%",
                "Media": round(mean, 4),
                "Desv. Estandar (S)": round(desv, 4),
                "Half-width": round(hval, 4),
                "Rango IC": round(2 * hval, 4),
            })

df_resumen = pd.DataFrame(resumen)
df_resumen
```

```python
fila_min_std = df_resumen.loc[df_resumen["Desv. Estandar (S)"].idxmin()]
fila_min_rango = df_resumen.loc[df_resumen["Rango IC"].idxmin()]

print("Combinacion con MENOR desviacion estandar:")
print(fila_min_std)
print("\nCombinacion con MENOR rango de intervalo de confianza:")
print(fila_min_rango)
```
